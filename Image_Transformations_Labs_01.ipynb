## SetupRun this cell first — installs/imports libraries and loads the input image.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image_path = "/content/437569044_449925474254730_265351335482707757_n.jpg"

image = cv2.imread(image_path) # Read image (OpenCV loads as BGR by default)

image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB) # Convert BGR to RGB for correct display with matplotlib

h, w = image.shape[:2]  # (height, width, channels)
print(f"Image loaded: {w}x{h}")
```

## Experiment 01: Translation of a 2D Image**Objective:** Perform image translation by shifting an image horizontally and vertically using an affine transformation matrix.

```python
# Translation values
tx = 100
ty = 50

# Translation matrix
M = np.float32([
    [1, 0, tx],
    [0, 1, ty]
])

# Apply transformation
translated = cv2.warpAffine(image, M, (w, h))

# Display
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)  # 1 row 2 columns, first position
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")
plt.subplot(1, 2, 2)
plt.imshow(translated)
plt.title("Translated Image")
plt.axis("off")
plt.show()
```

## Experiment 02: Scaling of a 2D Image**Objective:** Resize an image by applying scaling factors along the horizontal and vertical axes.

```python
# Scaling factors
sx = 5.0
sy = 8.5

# Scaling matrix
M = np.float32([
    [sx, 0, 0],
    [0, sy, 0]
])

scaled = cv2.warpAffine(
    image,
    M,
    (int(w * sx), int(h * sy))
)

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")
plt.subplot(1, 2, 2)
plt.imshow(scaled)
plt.title("Scaled Image")
plt.axis("off")
plt.show()
```

## Experiment 03: Rotation of a 2D Image**Objective:** Rotate an image around its center by a specified angle.

```python
# Center of image
center = (w // 2, h // 2)
# Rotation angle
angle = 45

# Rotation matrix
M = cv2.getRotationMatrix2D(center, angle, 1.0)
rotated = cv2.warpAffine(
    image,
    M,
    (w, h)
)

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")
plt.subplot(1, 2, 2)
plt.imshow(rotated)
plt.title("Rotated Image - 45°")
plt.axis("off")
plt.show()
```

## Experiment 04: Reflection About the X-Axis**Objective:** Perform image reflection across the horizontal axis (vertical flip).

```python
# Reflection about X-axis
reflected_x = cv2.flip(image, 0)

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")
plt.subplot(1, 2, 2)
plt.imshow(reflected_x)
plt.title("Reflection about X-axis")
plt.axis("off")
plt.show()
```

## Experiment 05: Reflection About the Y-Axis**Objective:** Create a mirror image by reflecting an image across the vertical axis (horizontal flip).

```python
# Reflection about Y-axis
reflected_y = cv2.flip(image, 1)

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")
plt.subplot(1, 2, 2)
plt.imshow(reflected_y)
plt.title("Reflection about Y-axis")
plt.axis("off")
plt.show()
```
